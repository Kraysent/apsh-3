#### Числа с единицами измерения

В 1999 NASA потеряло [Mars Climate Orbiter](https://ru.wikipedia.org/wiki/Mars_Climate_Orbiter#%D0%A5%D1%80%D0%BE%D0%BD%D0%BE%D0%BB%D0%BE%D0%B3%D0%B8%D1%8F_%D0%BF%D0%BE%D0%BB%D0%B5%D1%82%D0%B0_%D0%B8_%D1%80%D0%B5%D0%B7%D1%83%D0%BB%D1%8C%D1%82%D0%B0%D1%82%D1%8B). Компания Lockheed Martin, выпускавшая ПО для наземных систем, используемых аппаратом, выдавала расчёты в [фунт-силах](https://ru.wikipedia.org/wiki/%D0%A4%D1%83%D0%BD%D1%82_(%D0%B5%D0%B4%D0%B8%D0%BD%D0%B8%D1%86%D0%B0_%D0%B8%D0%B7%D0%BC%D0%B5%D1%80%D0%B5%D0%BD%D0%B8%D1%8F)). ПО NASA на самом аппарате, в свою очередь, использовало эти результаты для расчёта орбиты, но предполагало, что число было в ньютон-секундах. В результате орбита аппарата проходила в ближайшей точке на высоте 57 км над поверхностью Марса вместо расчитанных 80 км, что привело к более высокому атмосферному трению, что, в свою очередь, привело к разбитию аппарата о поверхность Марса.

![image.png](image.png)

Можно достаточно просто продемонстрировать, как такое случилось:

```python
thrust_force = 112.0   # тяга двигателя, фунт-силы (lbf)
burn_time = 4.5        # сколько секунд работал двигатель, с

thrust_impulse = thrust_force * burn_time  # импульс = сила * время

dry_mass = 588.0        # "сухая" масса аппарата, кг
propellant_mass = 52.0  # масса оставшегося топлива, кг

mass = dry_mass + propellant_mass  # полная масса = сухая + топливо

delta_v = thrust_impulse / mass  # dv = импульс / масса
print("Прирост скорости:", delta_v)  # 0.7875 фунт-сила * секунда / кг
```

Код написан, код выполняется, 0.78 выглядит как адекватное значение для м/с. Тем не менее, если бы мы не писали комментарии с единицами измерения, это значение бы потерялось уже на 3 строчке.

У комментариев как способа понимать единицы измерения в коде есть ворох проблем:
- Нет никакого кода, который бы осуществлял их проверку - правильность единиц измерения (не пытаемся ли мы сложить метры и килограммы?) остаётся на совести автора
- Комментарии устаревают - если где-то в коде выше килограммы поменяются на граммы, нам остаётся надеяться что программист не забудет поделить на 1000 во всех местах
- Если используются величина из другого модуля, нужно идти и смотреть в документации, какие величины возвращаются из функций этого модуля

Для того, чтобы избежать этих приключений, есть модуль `astropy.units`, который позволяет работать с величинами, к которым прикреплены единицы измерения. Делать их очень просто:

```python
from astropy import units as u

distance = 5 * u.m
speed = 10 * u.cm / u.s

time = distance / speed

print(time)
```

Единицы измерения никуда не потерялись (обратите внимание на то, что распечаталось). Теперь, если нужно привести к каким-нибудь человеческим единицам (вместо м * с / см), можно сделать `.to()`:

```python
print(time.to(u.s))
```

Посчитаем наш пример с импульсом выше:

```python
from astropy import units as u

thrust_force = 112.0 * u.imperial.lbf # для фунтов нужно использовать вложенный модуль imperial
burn_time = 4.5 * u.second # u.second и u.s - одно и то же

thrust_impulse = thrust_force * burn_time

dry_mass = 588.0 * u.kg
propellant_mass = 52.0 * u.kg

mass = dry_mass + propellant_mass

delta_v = thrust_impulse / mass
print("Прирост скорости:", delta_v, "==", delta_v.to(u.m / u.s))
```

Из примера выше видно, что можно делать произвольные математические операции, но стоит это явным образом проговорить:

```python
from astropy import units as u
import numpy as np

m = 50 * u.kg
d = 1200 * u.m

# деление на число:
print(m / 10)

# сложение, умножение, деление и тд:
print(m * d)
print(m / d)
print(m * m * d / m)

# возведение в степень:
print(m ** 2) # обратите внимание на единицу измерения - она тоже теперь в квадрате
print(np.sqrt(m)) # numpy очень дружит с astropy!
```

Astropy так же даст вам по рукам если вы попытаетесь сделать что-то странное:

```python
print(m + d)
```

```python
print(np.log10(m))
```

```python
print(np.sin(d))
```

#### Физические константы

astropy содержит много разных полезностей. Среди прочего - константы, на которые заранее навешаны единицы измерения:

```python
from astropy import constants

z = 0.0005

v = constants.c * z

print(v)
```

Список констант в astropy нескончаемый, в частности:

```python
print(constants.G)
print()
print(constants.k_B)
print()
print(constants.h)
print()

# частые физические величины:
print(constants.M_earth)
print()
print(constants.M_sun)
print()
print(constants.L_sun)
print()
print(constants.R_jup)
print()
```

Чаще всего, если вам нужна какая-то константа, можно попробовать:
- написать в коде `constants.{my_const}` и надеяться, что она есть - чаще всего так и будет
- загуглить `astropy my const` и воспользоваться документацией

#### Массивы с единицами измерения

Так как мы часто обрабатываем большие массивы данных, хочется уметь использовать `numpy` для этих вычислений, но одновременно пользоваться преимуществами единиц измерения в numpy. На наше счастье, astropy для вычислений внутри себя использует numpy, что значит, что мы можем пользоваться всеми преимуществами обоих одновременно!

Так же, как мы создавали числа с единицами измерения, можно создавать массивы с единицами измерения:

```python
import astropy.units as u

masses = [50, 45, 13] * u.M_earth

print(masses) # внутри это сразу становится numpy-массивом со всеми следствиями из этого:

print(masses / 2)

fractions = masses / u.M_sun
print(fractions.to(u.dimensionless_unscaled)) # так как величина fraction не имеет размерности (почему?), её можно привести к таковой при помощи константы dimensionless_unscaled
```

Если у вас уже есть numpy-массив, его можно так же привести к величинам с единицами измерения:

```python
import numpy as np
import astropy.units as u

values = np.array([100, 500, 1000, 5000])

values_with_units = values * u.kg
print(values_with_units)
```

или pandas-столбец:

```python
import pandas as pd

df = pd.DataFrame()
df["mass"] = np.random.uniform(-50, 50, 100)

values = df["mass"].values * (u.m / u.s)
print(values.to(u.km / u.s))
```

### Хранение таблиц и картинок

До сего момента мы в основном работали с данными, читая их из файла и делая какие-то преобразования над ними. В этой секции поговорим про то, как астрономы эти данные сохраняют.

Один из самых популярных форматов данных (вы с ним уже работали на практикумах) - FITS.

Структуру формата мы разберём ниже, но основная идея простая - хочется хранить несколько таблиц или картинок вместе, к каждой привязать метаданные (например, параметры телескопа, на который было снято изображение, ссылка на источник табличных данных) и при этом делать это всё эффективно.

#### Структура FITS

FITS состоит из набора страниц, называемых HDU - header data unit. Концептуально HDU - это то же самое, что страницы в Excel. В одном файле может находиться несколько HDU, каждый из которых содержит набор текстовых метаданных (headers) и (текстовых или бинарных) данных (data).

Первая страница FITS (первый HDU) ещё называется основной или primary.

![fits.png](fits.png)

Данные внутри каждой страницы, строго говоря, могут быть произвольными - таблица, картинка, текст или любой другой набор. Чаще всего в FITS-файлах встречаются таблицы или картинки.

#### Хранение таблиц


В файле II_299_gd.fits вы найдёте пример FITS-файла с текстовой таблицей внутри. Этот файл можно открыть любым текстовым редактором и посмотреть на него глазами - вы увидите, как расположены метаданные и табличные данные. Мы, однако, будем его читать через astropy.

```python
from astropy.io import fits

# открываем файл. Возвращается объект, содержащий список HDU (страниц) внутри файла.
hdu_list = fits.open("II_299_gd.fits")

print(hdu_list.info())
```

Из `hdu_list.info()` видно, что в файле два HDU: пустой `PRIMARY` (только метаданные, данных нет) и табличный HDU `gd.dat` с 1713 строками и 13 колонками. Достанем нужный HDU по индексу и посмотрим его заголовок:

```python
primary_hdu = hdu_list[0]
header = dict(primary_hdu.header)
print(header.keys())
print(header)
```

`primary_hdu.header` - это словарь, содержащий метаданные, привязанные к этой странице. В данном случае это размерности данных, абстракт, ссылка на источник, авторы, дата и тд.

Чаще всего primary HDU содержит только метаданные об источнике без настоящих данных. Как мы увидели выше, в этом же файле есть второй HDU - посмотрим на него.

```python
second_hdu = hdu_list[1]
header = dict(second_hdu.header)

print(header)
```

Несколько интересных нам вещей:
- `'XTENSION': 'TABLE'` - говорит о том, что в этом HDU лежит таблица
- `'NAXIS1': 76` - в таблице 76 колонок
- `'NAXIS2': 1713` - в таблице 1713 строк

Все остальные заголовки - это другая (иногда) полезная информация о таблице.

Посмотрим на саму таблицу:

```python
import pandas as pd

tbl = pd.DataFrame(second_hdu.data)
tbl
```

Эти данные преобразуются к pandas-таблице одной строкой. Это значит, что мы теперь можем работать с этими данными как угодно - преобразовывать к numpy-массивам, совершать математические операции и так далее.
