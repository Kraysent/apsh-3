В ноутбуке по numpy мы уже делали подобные операции. Код для чтения той же таблицы в numpy будет выглядеть примерно так:

````python
import numpy as np

data = np.genfromtxt("stars.csv", delimiter=",", dtype=None, names=True)

target_stars = ["Pollux", "Bellatrix", "Dubhe", "Betelgeuse"]

selected_rows = [row for row in data if row[4] in target_stars]

for row in selected_rows:
    for i in range(len(data.dtype.names)):
        name = data.dtype.names[i]
        value = row[i]
        print(f"{name}: {value}")
   
    print()
````

В этом ноутбуке мы познакомимся с ещё одним способом работы с таблицами - библиотекой `pandas`. По сравнению с `numpy` она даёт сильно большую гибкость для перекладывания табличных данных из одного места в другое.

Сразу к примерам - чтение нашей таблицы:

````python
import pandas as pd

df = pd.read_csv("stars.csv")

print(df)
````

Выглядит даже проще, чем `numpy` (который уже выглядел достаточно просто) и создаёт нам основной объект таблицы в `pandas` - `DataFrame`.

Pandas оперирует таблицами, объект которых называется `DataFrame` - "каркас данных". Он, в свою очередь, содержит столбцы, которые называются `Series`.

Изучим пристальнее таблицу сверху:

````python
print(type(df)) # <-- DataFrame

print(df["name"]) # Получаем столбец

print(type(df["name"])) # <-- Series
````

Оба этих объекта часто встречаются в ошибках, поэтому полезно помнить их названия.

После чтения этих таблиц, о них можно получать разную информацию. В частности:

````python
print(df.head()) # Выводит первые 5 строк таблицы
print()

print(df.tail(n=3)) # Выводит последние n строк таблицы
print()

df.info() # Выводит информацию о столбцах таблицы
````

Первая сила pandas в гибкости чтения данных - он умеет создавать таблицы из CSV, TSV, Excel и всего остального, что можно себе представить. Список поддерживаемых форматов: https://pandas.pydata.org/docs/reference/io.html

Имейте ввиду, что для чтения из excel нужно установить библиотеку `openpyxl` в дополнение к самому `pandas` (`pip install openpyxl` - он уже установлен здесь).

````python
tsv_df = pd.read_csv("stars.tsv", sep="\t") # как CSV, но разделитель - знак табуляции
print(tsv_df.head())
print()

excel_df = pd.read_excel("stars.xlsx")
print(excel_df.head())
````
