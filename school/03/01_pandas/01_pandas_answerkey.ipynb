### Опять таблицы

На прошлых школах мы касались табличных данных и что с ними можно делать. На этой мы поговорим подробнее о разных видах таблиц и какие средства в Python есть для работы с ними.

Сначала вспомним, как читать таблицы.

<div class="alert alert-block alert-warning" style="margin-top: 20px">

<font size=4>**Задание 1**</font>

Встроенными средствами в Python прочитать таблицу `stars.csv`. Вывести все параметры в любом удобном виде для звёзд Pollux, Bellatrix, Dubhe и Betelgeuse.

</div>

````python
target_stars = ["Pollux", "Bellatrix", "Dubhe", "Betelgeuse"]

with open("stars.csv") as f:
    lines = f.read().splitlines()

header = lines[0].split(",")
for line in lines[1:]:
    values = line.split(",")
    if values[4] in target_stars:
        for i in range(len(header)):
            print(f"{header[i]}: {values[i]}")
        print()
````

В ноутбуке по numpy мы уже делали подобные операции. Код для чтения той же таблицы в numpy будет выглядеть примерно так:

````python
import numpy as np

data = np.genfromtxt("stars.csv", delimiter=",", dtype=None, names=True)

target_stars = ["Pollux", "Bellatrix", "Dubhe", "Betelgeuse"]

selected_rows = [row for row in data if row[4] in target_stars]

for row in selected_rows:
    for i in range(len(data.dtype.names)):
        name = data.dtype.names[i]
        value = row[i]
        print(f"{name}: {value}")
   
    print()
````

В этом ноутбуке мы познакомимся с ещё одним способом работы с таблицами - библиотекой `pandas`. По сравнению с `numpy` она даёт сильно большую гибкость для перекладывания табличных данных из одного места в другое.

### Что с ними делать?

Теперь будем говорить о том, зачем это всё надо.

#### Получение данных о таблице

````python
import pandas as pd

df = pd.read_csv("stars.csv")

print(df.shape) # кортеж, отдающий число строк (первое значение) и столбцов (второе значение) в таблице

# Это значит, что отдельно число строк и столбцов можно получить простой индексацией:
print("Число строк:", df.shape[0])
print("Число столбцов:", df.shape[1])

print("Число строк, но проще:", len(df))
print("Число столбцов, но проще:", len(df.columns))

print(df.dtypes)  # вывод типов данных для всех столбцов
````

#### Получение данных

````python
names = list(df["name"]) # получение списка всех данных в столбце

# names - это обычный список, над которым можно делать все знакомые нам операции
print(names)
print(names[10])
print(names[5:10])
````

#### Математика

Внутри себя pandas использует numpy в качестве хранилища массивов данных. Это значит, что все знакомые математические операции из numpy можно таким же образом делать над столбцами:

````python
import numpy as np

values = df["apparent_magnitude"]

mag_sq = values ** 2
print(list(mag_sq))
print(list(values - 10))
print(list(np.log10(2.5 ** values)))
````

#### Изменение таблиц

Функциональность, которой нет в numpy - это добавление новых колонок в существующую таблицу:

````python
import numpy as np

df_copy = df

df_copy["catalog"] = "Hipparcos" #     можно добавить константное значение
print(list(df_copy["catalog"])) # которое само размножится на число строк в таблице

df_copy["distance"] = np.random.random(len(df_copy)) * 100 # можно записать произвольный numpy-массив
print(list(df_copy["distance"]))

# можно скопировать другой стобец
df_copy["distance_copy"] = df_copy["distance"]
print(list(df_copy["distance_copy"]))

# или делать математические операции над другими стобцами
df_copy["abs_magnitude"] = df_copy["apparent_magnitude"] - 5 * (np.log10(df_copy["distance"]) - 1)
print(list(df_copy["abs_magnitude"]))
````

Вероятные ошибки, достойные примечания (выяснить, что именно не так, остаётся в качестве умственного упражнения):

````python
df_copy["new_distance"] = np.random.random()
print(list(df_copy["new_distance"]))

df_copy["something"] = np.zeros(shape=(10,))
````

Таблицы можно так же собирать с нуля (или из частей других таблиц):

````python
df2 = pd.DataFrame()

df2["column_1"] = np.ones(shape=(15,))
df2["column_2"] = df2["column_1"] + df["distance"][0:len(df2)]

print(df2)
````

#### Математика, но не над числами

Так как pandas умеет хранить не только численные данные, мы можем делать и другие операции, которые в numpy были бы не так просты:

````python
import pandas as pd

tbl = pd.DataFrame()

tbl["name"] = df["name"]
tbl["id"] = df["hip_id"]
tbl["prefixed_name"] = "HIP (" + tbl["id"].astype(str) + ") " + df["name"] # конкатенация строк

# .astype говорит, что все значения столбца надо преобразовать к строковому типу

print(tbl)
````

#### Вырезание столбцов

Если нам нужна только часть таблицы, есть более простая опция создания такой "подтаблицы", чем создание DataFrame с нуля:

````python
tbl2 = df[["name", "apparent_magnitude"]]

print(tbl2.head())
````

Что происходит?

```python
tbl2 = df[["name", "apparent_magnitude"]]
```

Мы уже видели передачу чисел в квадратные скобки:
```python
lst = [10, 20, 30]
print(lst[2])
```

и строк:
```python
print(df["name"])
```

Строго говоря, в квадратные скобки можно передавать какой угодно объект, если известно, как его интерпретировать. В данном случае мы передаём в DataFrame список строк. Pandas знает, что если передан список строк, нужно сделать таблицу из колонок, имена которых перечислены в этом списке.


#### Фильтрация

Так же как и numpy, pandas умеет фильтровать данные. Примеры:

````python
my_filter = [True] * (len(tbl2) // 2 + 1) + [False] * (len(tbl2) // 2) # а что тут происходит?
print(my_filter)
print("-" * 50)

# можно фильтровать произвольным списком булевых значений:
print(tbl2[my_filter])
print("-" * 50)

# а можно фильтровать и на основании других столбцов:
print(tbl2[tbl2["name"].str.startswith("B")])
print("-" * 50)
print(tbl2[tbl2["apparent_magnitude"] > 2])
````

В таблице `🫠.table` найти объект, который был обнаружен раньше всего, и вывести список имён объектов, которые имеют видимую звёздную величину, большую, чем этот объект.

````python
import pandas as pd

with open("🫠.table") as f:
    lines = f.read().splitlines()

with open("stars_cleaned.csv", "w") as out:
    for line in lines:
        line = line.strip()
        if len(line) >= 4 and line[:2] == ">>" and line[-2:] == "<<":
            parts = line[2:-2].strip().split(":::")
            out.write(",".join(parts) + "\n")

df = pd.read_csv("stars_cleaned.csv")

min_date = min(list(df["discovery_date"]))
earliest_mag = list(df[df["discovery_date"] == min_date]["apparent_magnitude"])[0]
print(list(df[df["apparent_magnitude"] > earliest_mag]["name"]))
````

### Уроки рисования

Так же, как и с numpy-массивами, столбцы pandas-таблиц можно рисовать на графиках:

````python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame()
df["star"] = ["Sirius", "Canopus", "Arcturus", "Vega", "Capella"]
df["apparent_magnitude"] = [-1.46, -0.74, -0.05, 0.03, 0.08]
print(df)

plt.figure()
plt.bar(df["star"], df["apparent_magnitude"])
plt.gca().invert_yaxis()
plt.show()
````
